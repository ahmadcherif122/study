# 2.6 Simplification des classes avec `dataclass`

## 2.6.1 Définition d’une `dataclass`

````python
from dataclasses import dataclass

@dataclass
class Etudiant:
    nom: str
    age: int
````

Une `dataclass` permet de définir une classe de données avec moins de code.


## 2.6.2 Le décorateur `@dataclass`

`@dataclass` indique à Python que cette classe doit être traitée comme une dataclass.

Il permet notamment à Python de générer automatiquement certaines méthodes courantes, comme le constructeur `__init__()`.

On place `@dataclass` juste avant la définition de la classe.


````python
from dataclasses import dataclass

@dataclass
class Etudiant:
    nom: str
    age: int

e = Etudiant("Ali", 20)

print(e.nom)
print(e.age)
````

## 2.6.3 Classe classique vs `dataclass`

### Classe classique

Sans `dataclass`, il faut écrire explicitement le constructeur :


````python
class Etudiant:
    def __init__(self, nom, age, moyenne=0.0):
        self.nom = nom
        self.age = age
        self.moyenne = moyenne
````

### Avec `dataclass`


````python
from dataclasses import dataclass

@dataclass
class Etudiant:
    nom: str
    age: int
    moyenne: float = 0.0
````

| Classe classique | `dataclass` |
|---|---|
| `__init__()` écrit manuellement | `__init__()` généré automatiquement |
| Plus de code | Moins de code |
| Attributs initialisés dans `__init__()` | Attributs déclarés directement |
| Plus répétitif | Plus lisible |

Une `dataclass` permet également de définir une valeur par défaut pour un attribut.


## 2.6.4 Classe classique vs `dataclass` — `ClassVar`

`nombre_etudiants` est une variable de classe et ne doit pas être considérée comme un attribut de l’objet par `dataclass`.


````python
from dataclasses import dataclass
from typing import ClassVar

@dataclass
class Etudiant:
    nombre_etudiants: ClassVar[int] = 0

    nom: str
    age: int
    moyenne: float

e = Etudiant("Ahmed", 20, 15.5)

print(Etudiant.nombre_etudiants)
````

## 2.6.5 Classe classique vs `dataclass` — `__post_init__`


````python
class Etudiant:
    # Variable de classe
    nombre_etudiants = 0

    def __init__(self, nom, age, moyenne):
        # Variables d'instance
        self.nom = nom
        self.age = age
        self.moyenne = moyenne

        # Incrémentation du compteur
        Etudiant.nombre_etudiants += 1

e1 = Etudiant("Ali", 20, 15.5)
e2 = Etudiant("Sara", 21, 17.2)

print(e1.nom, e1.age, e1.moyenne)
print("Nombre total d'étudiants :", Etudiant.nombre_etudiants)
````

### Avec `dataclass`


````python
from dataclasses import dataclass
from typing import ClassVar

@dataclass
class Etudiant:
    nombre_etudiants: ClassVar[int] = 0

    nom: str
    age: int
    moyenne: float

    def __post_init__(self):
        Etudiant.nombre_etudiants += 1

e1 = Etudiant("Ali", 20, 15.5)
e2 = Etudiant("Sara", 21, 17.2)

print(e1)
print("Nombre total d'étudiants :", Etudiant.nombre_etudiants)
````

## 2.6.6 Attributs typés : annotations de type (*Type Hints*)

Dans une `dataclass`, les attributs sont déclarés à l'aide d'annotations de type (*Type Hints*), qui permettent à Python d'identifier les champs de la classe.

Les *Type Hints* indiquent le type attendu d'une donnée, mais ne constituent pas à eux seuls une validation stricte à l'exécution.


````python
@dataclass
class Etudiant:
    nom: str
    age: int
    moyenne: float
````

`age: int` signifie que `age` est prévu pour être un entier.

| Attribut | Type attendu |
|---|---|
| `nom` | `str` |
| `age` | `int` |
| `moyenne` | `float` |


Les annotations sont particulièrement utiles parce que `@dataclass` les utilise pour savoir quels champs appartiennent à la classe et pour générer automatiquement certaines méthodes, notamment `__init__`.


## 2.6.7 Les types complexes dans `dataclass`

Les champs d'une `dataclass` peuvent être de types simples (`int`, `float`, `str`), mais aussi de types complexes comme les listes (`list`), les dictionnaires (`dict`), les ensembles (`set`), les tuples (`tuple`) ou des objets d'autres classes.

Pour déclarer ces champs, on utilise les type hints afin de préciser les types attendus. Pour les collections mutables utilisées comme valeurs par défaut, on utilise `field(default_factory=...)` afin de créer une nouvelle collection pour chaque instance et d'éviter le partage involontaire de données entre objets.

````python
from dataclasses import dataclass, field
from typing import ClassVar

@dataclass
class Adresse:
    ville: str
    code_postal: str

@dataclass
class Etudiant:
    nombre_etudiants: ClassVar[int] = 0

    nom: str
    age: int
    moyenne: float

    # Attribut de type objet
    adresse: Adresse

    # Attributs de types complexes
    notes: list[float] = field(default_factory=list)
    competences: dict[str, int] = field(default_factory=dict)
    absences: set[str] = field(default_factory=set)

    def __post_init__(self):
        Etudiant.nombre_etudiants += 1

a = Adresse("Sfax", "3000")
b = Adresse("Tunis", "2000")
e1 = Etudiant("Ahmed", 20, 15.5,a)
e2 = Etudiant("Sarra", 21, 16.0,b)

e1.notes.append(17.0)
e1.notes.append(14.5)

e1.competences["Python"] = 3
e1.competences["Java"] = 2

e1.absences.add("10/10/2026")

print(e1)
print(e2)

print("Notes Ahmed :", e1.notes)
print("Notes Sarra :", e2.notes)

print("Nombre d'étudiants :", Etudiant.nombre_etudiants)
````
