## 1. Méthodes d’instance → self

```python
class Etudiant:
    def __init__(self, nom):
        self.nom = nom

    def afficher(self):
        print("Nom :", self.nom)
        

#Utilisation de la méthode d'instance   

#On crée deux étudiants :
e1 = Etudiant("Ahmed")
e2 = Etudiant("Sarra")


#Quand on écrit :
e1.afficher()  #Python sait que la méthode doit travailler sur e1 => self=e1

e2.afficher()  #self=e2
```

**Méthode d'instance = une action qui concerne un objet particulier**

##### utiliser @property pour accéder à un attribut privé avec une syntaxe similaire à celle d'un attribut classique.

```python
class CompteBancaire:
    def __init__(self, solde):
        self.__solde = solde

    @property
    def solde(self):
        return self.__solde

    @solde.setter
    def solde(self, s):
        if s >= 0:
            self.__solde = s
        else:
            print("Solde invalide")
```

```python
c = CompteBancaire(1000)

print(c.solde)

c.solde = 2000
print(c.solde)

c.solde = -500
print(c.solde)
```

#### @property permet de définir une méthode qui est utilisée comme un attribut.

#### @nom.setter permet de contrôler la modification de cette propriété.
