## **Objectif : trier la liste L suivant les prix des fruits**

```python
class Fruit():
    def __init__(self, name, price):
        self.name = name
        self.price = price
        
L = [Fruit("Cherry",10), Fruit("Apple",5), Fruit("Blueberry",20)]

for f in sorted(L, key=lambda x:x.price):
    print(f.name)
```

```python

class Fruit():
    def __init__(self, name, price):
        self.name = name
        self.price = price
    
    def sort_priority(self):
        return self.price
    

L = [Fruit("Cherry",10), Fruit("Apple",5), Fruit("Blueberry",20)]

print("---triés par prix, faisant référence à une méthde de classe---")
for f in sorted(L, key=Fruit.sort_priority):
    print(f.name)
    
print("---Une autre façon de faire la meme chose!!---")
for f in sorted(L, key=lambda x: x.sort_priority()):
    print(f.name)
```
